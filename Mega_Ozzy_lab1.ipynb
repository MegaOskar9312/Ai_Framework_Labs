```python
import pandas as pd

url = 'https://docs.google.com/uc?export=download&id=1cq9JDz8-zwzoGL6udiT_ogHHRWdnsd1q'
data = pd.read_csv(url)

print('Размер таблицы:', data.shape)
display(data.head())
```

1

```python
reason_counts = data['reason'].value_counts()

print(reason_counts)
print('cамая частая причина:', reason_counts.idxmax())
```

2

```python
no_education = data[
    (data['Medu'] == 0) & (data['Fedu'] == 0)
]

print('количество студентов:', len(no_education))
```

3

```python
min_age = data.loc[data['school'] == 'MS', 'age'].min()

print('минимальный возраст:', min_age)
```

4

```python
odd_absences = data[data['absences'] % 2 == 1]

print('количество студентов:', len(odd_absences))
```
